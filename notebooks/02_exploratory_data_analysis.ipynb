```python
# ==============================================================================
# WEEK 2: EXPLORATORY DATA ANALYSIS (EDA)
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Create visualizations directory if it doesn't exist
os.makedirs('../visualizations', exist_ok=True)

# 1. Load Clean Dataset
df = pd.read_csv('../data/processed/cleaned_food_delivery.csv')
print(f"Clean Dataset Loaded: {df.shape[0]} rows, {df.shape[1]} columns.\n")

# 2. Delivery-Duration Baseline Statistics
duration_col = 'Time_taken (min)'
s = df[duration_col]

baseline_stats = pd.DataFrame({
    'Statistic': [
        'Count', 'Mean', 'Median', 'Min', 
        'Max', '25th Percentile', '75th Percentile', 'Std Dev'
    ],
    'Value': [
        s.count(),
        s.mean(),
        s.median(),
        s.min(),
        s.max(),
        np.percentile(s, 25),
        np.percentile(s, 75),
        s.std()
    ]
})

print("--- Delivery Duration Baseline Statistics ---")
print(baseline_stats.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Ensure the visualizations directory exists
os.makedirs('../visualizations', exist_ok=True)

# Generate scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Distance_km', y='Time_taken (min)', alpha=0.3, color='royalblue')
plt.title('Initial Scatter Plot: Delivery Distance vs. Delivery Duration')
plt.xlabel('Delivery Distance (km)')
plt.ylabel('Delivery Duration (minutes)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Save to the required path
plt.savefig('../visualizations/initial_distance_vs_duration.png', dpi=300)
plt.show()

print("Scatter plot successfully saved to visualizations/initial_distance_vs_duration.png")
```
