```python
# ==============================================================================
# WEEK 3: VISUALIZATION & INTERPRETATION
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Create visualizations directory if it doesn't exist
os.makedirs('../visualizations', exist_ok=True)

# Load cleaned dataset
df = pd.read_csv('../data/processed/cleaned_food_delivery.csv')
print(f"Loaded Clean Dataset: {df.shape[0]} rows, {df.shape[1]} columns.\n")

# Set global plotting theme
sns.set_theme(style="whitegrid")

# ------------------------------------------------------------------------------
# Chart 1: Delivery Duration Distribution (Histogram)
# ------------------------------------------------------------------------------
plt.figure(figsize=(9, 6))
sns.histplot(df['Time_taken (min)'], bins=30, kde=True, color='teal')
plt.title('01. Distribution of Delivery Durations', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Delivery Duration (minutes)', fontsize=12)
plt.ylabel('Number of Deliveries', fontsize=12)
plt.tight_layout()
plt.savefig('../visualizations/01_delivery_duration_distribution.png', dpi=300)
plt.close()

# ------------------------------------------------------------------------------
# Chart 2: Traffic Comparison (Bar Chart of Median Duration)
# ------------------------------------------------------------------------------
traffic_median = df.groupby('Road_traffic_density')['Time_taken (min)'].median().reset_index()
traffic_median = traffic_median.sort_values(by='Time_taken (min)', ascending=False)

plt.figure(figsize=(9, 6))
sns.barplot(data=traffic_median, x='Road_traffic_density', y='Time_taken (min)', palette='Blues_r')
plt.title('02. Median Delivery Duration by Road Traffic Density', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Road Traffic Density', fontsize=12)
plt.ylabel('Median Delivery Duration (minutes)', fontsize=12)
plt.tight_layout()
plt.savefig('../visualizations/02_traffic_comparison.png', dpi=300)
plt.close()

# ------------------------------------------------------------------------------
# Chart 3: Weather Comparison (Bar Chart of Average Duration)
# ------------------------------------------------------------------------------
weather_mean = df.groupby('Weather_conditions')['Time_taken (min)'].mean().reset_index()
weather_mean = weather_mean.sort_values(by='Time_taken (min)', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(data=weather_mean, x='Weather_conditions', y='Time_taken (min)', palette='viridis')
plt.title('03. Average Delivery Duration Across Weather Conditions', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Weather Conditions', fontsize=12)
plt.ylabel('Average Delivery Duration (minutes)', fontsize=12)
plt.xticks(rotation=25)
plt.tight_layout()
plt.savefig('../visualizations/03_weather_comparison.png', dpi=300)
plt.close()

# ------------------------------------------------------------------------------
# Chart 4: Distance vs. Delivery Duration (Scatter Plot)
# ------------------------------------------------------------------------------
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Distance_km', y='Time_taken (min)', alpha=0.3, color='royalblue')
plt.title('04. Delivery Distance vs. Delivery Duration', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Delivery Distance (km)', fontsize=12)
plt.ylabel('Delivery Duration (minutes)', fontsize=12)
plt.tight_layout()
plt.savefig('../visualizations/04_distance_vs_duration.png', dpi=300)
plt.close()

# ------------------------------------------------------------------------------
# Chart 5: Vehicle Comparison (Box Plot)
# ------------------------------------------------------------------------------
plt.figure(figsize=(9, 6))
sns.boxplot(data=df, x='Type_of_vehicle', y='Time_taken (min)', palette='Set2')
plt.title('05. Delivery Duration Distribution Across Vehicle Types', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Vehicle Type', fontsize=12)
plt.ylabel('Delivery Duration (minutes)', fontsize=12)
plt.tight_layout()
plt.savefig('../visualizations/05_vehicle_comparison.png', dpi=300)
plt.close()

print("Success! All 5 finalized charts exported to visualizations/ folder.")
```
